```python
# df['SearchTerm'] = df['SearchTerm'].str.replace('associated food prompt', '', case=False, regex=False)
# df['SearchTerm'] = df['SearchTerm'].str.replace('compound food template', '', case=False, regex=False)
# Group by 3 columns
# group_cols = ['ID', 'RecallNumber', 'MealID']
group_cols = ['PPT_ID', 'RecallNumber', 'MealID']

# Columns to aggregate
agg_cols = ['SearchTerm', 'NameEn']

# Custom aggregation functions
def join_non_empty_comma(values):
    cleaned = values.dropna().astype(str)
    cleaned = cleaned[cleaned.str.strip() != '']
    return ','.join(cleaned)

def join_non_empty_space(values):
    cleaned = values.dropna().astype(str)
    cleaned = cleaned[cleaned.str.strip() != '']
    return ' '.join(cleaned)

# Apply different aggregation functions per column
agg_df = df.groupby(group_cols, as_index=False).agg({
    'NameEn': join_non_empty_comma,
    'SearchTerm': join_non_empty_comma
   #  'Lost foods': join_non_empty_comma
})

# Output to CSV or view it
print(agg_df)
agg_df.to_csv('/home/dmt-adm/missing_food/Y17_MF1L_withAFP_for_Tom_grouped.csv', index=False)
```
